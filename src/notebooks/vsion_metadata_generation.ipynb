```python
import json 
import re
```

```python
text_data_path = '../../data/json_data/chunks_with_metadata.json'
vision_data_path = '../../data/json_data/generated_image_captions_v2.json'
```

```python
with open(text_data_path, 'r') as file : 
        text_data = json.load(file)                
with open(vision_data_path, 'r') as file:
        vision_data = json.load(file)
```

```python
lookup_fig = {item["title"]: item for item in text_data}
```

```python
lookup_fig 
```

```python
new_json_output = []
for item in vision_data:
    key =  re.search(r"Figure\s\d+-\d+", item["original_tag"]).group(0)
    item_b = lookup_fig.get(key)
    print(item_b)
    if item_b:
        # Build the new combined record
        combined_item = {
            "title": key,
            "content": item.get("caption")[0],
            "metadata": {
                    "chapter": item_b.get("metadata", {}).get("chapter", ""),
                    "section": item_b.get("metadata", {}).get("section", ""),
                    "subsection": item_b.get("metadata", {}).get("subsection", ""),
                    "is_figure": item_b.get("metadata", {}).get("is_figure", False),
                    "is_table": item_b.get("metadata", {}).get("is_table", False),
                    "info_source": item.get("info_source", ""),
        }
        }
        new_json_output.append(combined_item)
```

```python
new_json_output
```

```python
with open("../../data/json_data/vision_data_metadata.json", "w", encoding="utf-8") as f:
    json.dump(new_json_output, f, ensure_ascii=False, indent=2)
```
