```python
def add_metadata(row):
    if row['color'] == 0 and row['size'] == 18.602558135986328 and row['font'] == 'ACaslonPro-Bold':
        return 'chapter'
    elif row['color'] == 9905811 and row['size'] == 10.20140266418457 and row['font'] == 'ACaslonPro-Bold':
        return 'section'
    elif row['color'] == 14632002 and row['size'] == 11.401566505432129 and row['font'] == 'ACaslonPro-Bold':
        return 'subsection'
    elif row['color'] == 13422054 and row['size'] == 10.20140266418457 and row['font'] == 'ACaslonPro-Bold':
        return 'table'
    elif row['color'] == 14632002 and row['size'] == 7.801072597503662 and row['font'] == 'ACaslonPro-Bold':
        return 'figure'
    else:
        return 'normale'
    
```

```python
# Apply the function and create a new column
df['label'] = df.apply(assign_label, axis=1)
```

```python
# Re-define the logic to build chunks
chunks = []
current_chunk = {
    "title": "",
    "content": "",
    "font": "",
    "color": "",
    "size": ""
}

for _, row in df.iterrows():
    if row['label'] == 'title':
        # Save the current chunk if it has content
        if current_chunk["title"] and current_chunk["content"]:
            chunks.append(current_chunk)
        # Start a new chunk
        current_chunk = {
            "title": row['text'].strip(),
            "content": "",
            "font": row['font'],
            "color": row['color'],
            "size": row['size']
        }
    else:
        current_chunk["content"] += row['text'].strip() + " "

# Add the last chunk if it's not empty
if current_chunk["title"] and current_chunk["content"]:
    chunks.append(current_chunk)

# Convert to DataFrame for visualization
chunks_df = pd.DataFrame(chunks)
```

```python
chunks_df
```

```python
chunks_df['type'] = chunks_df.apply(add_metadata, axis=1)
```

```python
# Initialize context variables
current_chapter = None
current_section = None
current_subsection = None

# List to hold enriched chunk data
chunks_with_metadata = []

# Iterate through the DataFrame rows
for _, row in chunks_df.iterrows():
    # Update context
    if row['type'] == 'chapter':
        current_chapter = row['title']
        current_section = None
        current_subsection = None
    elif row['type'] == 'section':
        current_section = row['title']
        current_subsection = None
    elif row['type'] == 'subsection':
        current_subsection = row['title']

    # Check if it's a figure or table
    is_figure = row['type'] == 'figure'
    is_table = row['type'] == 'table'

    # Build the metadata
    metadata = {
        "chapter": current_chapter,
        "section": current_section,
        "subsection": current_subsection,
        "is_figure": is_figure,
        "is_table": is_table
    }

    # Add the chunk with metadata
    chunks_with_metadata.append({
        "title": row['title'],
        "content": row['content'],
        "metadata": metadata
    })

# Save to a JSON file
json_path = "chunks_with_metadata.json"
with open(json_path, "w", encoding="utf-8") as f:
    json.dump(chunks_with_metadata, f, ensure_ascii=False, indent=2)
```
